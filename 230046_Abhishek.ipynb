#Question: 1.


```python
# Part a:
sentence = input("Write you sentence here: ")

# Part b:
print(sentence.upper())

# Part c:
vovels = "aeiouAEIOU"
count = 0

for ch in sentence:
  if ch in vovels:
    count = count+1

print("Number of vovels present in the sentence are ",count,".",sep="")

# Part d:
reversed_sentence = sentence[::-1]
print("Reversed sentence:\n",reversed_sentence)
```

#Question: 4.

```python
# Part a:
def process_data(data):
  if type(data) != list:
    return "Only list is allowed!"

  # Part b:
  sum = 0
  for ele in data:
    sum = sum + ele
  mean = sum/(len(data))
  print("The mean of the data:",mean)

  # Part c:
  new_list = [ele for ele in data if ele>mean]

  # Part d1:
  result = (mean,new_list)

  return result

# Part d2:
data = list(map(int,input("Write down your data enteries here:").split()))
print("The given data:",data)
print("Tuple returned from the function:",process_data(data))
```
